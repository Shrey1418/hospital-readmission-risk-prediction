```python
# Verify discharge_disposition_id labels against the official mapping
section = None
rows = {}
for line in open("data/IDS_mapping.csv", encoding="utf-8-sig"):
    line = line.strip()
    if not line:
        continue
    if line.endswith(",description"):
        section = line.split(",")[0]
        continue
    rows.setdefault(section, []).append(line)

for i in [1, 5, 11, 13, 14, 19, 20, 21, 22]:
    print([r for r in rows["discharge_disposition_id"] if r.startswith(f"{i},")])
```
